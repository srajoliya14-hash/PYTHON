```python
"""create an array from number 40 - 132
create an 4-d array of zeros 
create an 2-d array of 1's
cteate an array where the number 213 isrepeated 12 times 
create an array where the number 1.001 is repeated 8 times 
create an identity matrix of shape (4,4)
create an identity matrix of shape (9,9)
"""
```

```python
arr1 = np.arange(40, 133)
print(arr1)
 
arr1 = np.zeros((2, 2, 2, 2))
print(arr1)

arr1 = np.ones([2,2,])
print(arr1)
 
arr1 = np.full(12, 213)
print(arr1)
 
arr1 = np.full(8, 1.001)
print(arr1)
 
arr1 = np.eye(4)
print(arr1)
 
arr1 = np.eye(9)
print(arr1)
```

```python
# aggregate function
arr1 = np.arange(30,45)
print(arr1)
print(sum(arr1))
print(min(arr1))
print(max(arr1))
print(arr1.std()) # standard deviation
print(arr1.var()) # variance
print(arr1.argmax()) # argument maximum
print(arr1.argmin()) # argument mimimum
print(arr1.mean()) # avgrage
```
