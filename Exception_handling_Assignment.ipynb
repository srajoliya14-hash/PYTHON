```python
def divide(a, b):
    try:
        result = a / b
    except ZeroDivisionError:
        print("Cannot divide by zero")
    else:
        print("Result:", result)
    finally:
        print("Execution completed")


divide(10, 2)
```

```python
def convert_integer(value):
    try:
        num = int(value)
    except ValueError:
        print("Invalid input")
    else:
        print("Converted value:", num)
    finally:
        print("Conversion attempt finished")


convert_integer("25")
```

```python
def square(value):
    try:
        num = float(value)
        result = num ** 2
    except ValueError:
        print("Invalid input")
    else:
        print("Square:", result)
    finally:
        print("Process done")


square("5.5")
```

```python
def floor_division(a, b):
    try:
        result = a // b
    except ZeroDivisionError:
        print("Cannot divide by zero")
    else:
        print("Quotient:", result)
    finally:
        print("Function executed")


floor_division(10, 3)
```

```python
def reciprocal(value):
    try:
        num = float(value)
        result = 1 / num
    except ZeroDivisionError:
        print("Cannot find reciprocal of zero")
    except ValueError:
        print("Invalid input")
    else:
        print("Reciprocal:", result)
    finally:
        print("End of operation")


reciprocal(5)
```

```python
def multiply(a, b):
    try:
        num1 = int(a)
        num2 = int(b)
        result = num1 * num2
    except ValueError:
        print("Invalid input")
    else:
        print("Product:", result)
    finally:
        print("Multiplication tried")


multiply("5", "4")
```

```python
def percentage(marks, total):
    try:
        result = (marks / total) * 100
    except ZeroDivisionError:
        print("Total marks cannot be zero")
    else:
        print("Percentage:", result, "%")
    finally:
        print("Calculation finished")


percentage(450, 500)
```

```python
def half(value):
    try:
        num = float(value)
        result = num / 2
    except ValueError:
        print("Invalid input")
    else:
        print("Half:", result)
    finally:
        print("Conversion completed")


half("20.5")
```

```python
def modulus(a, b):
    try:
        result = a % b
    except ZeroDivisionError:
        print("Cannot find modulus with zero")
    else:
        print("Remainder:", result)
    finally:
        print("Modulus attempt done")


modulus(10, 3)
```

```python
def power(a, b):
    try:
        result = a ** b
    except Exception:
        print("Invalid input")
    else:
        print("Result:", result)
    finally:
        print("Power function executed")


power(2, 3)
```

```python
def absolute_value(value):
    try:
        num = float(value)
        result = abs(num)
    except ValueError:
        print("Invalid input")
    else:
        print("Absolute value:", result)
    finally:
        print("Absolute check completed")


absolute_value("-25")
```

```python
def divide_three(a, b, c):
    try:
        result1 = a / b
        result2 = result1 / c
    except ZeroDivisionError:
        print("Cannot divide by zero")
    else:
        print("Final result:", result2)
    finally:
        print("Division process ended")


divide_three(100, 5, 2)
```

```python
def double(value):
    try:
        num = int(value)
        result = num * 2
    except ValueError:
        print("Invalid input")
    else:
        print("Double:", result)
    finally:
        print("Closing message")


double("15")
```

```python
def subtract(a, b):
    try:
        num1 = float(a)
        num2 = float(b)
        result = num1 - num2
    except ValueError:
        print("Invalid input")
    else:
        print("Difference:", result)
    finally:
        print("Subtraction attempt finished")


subtract("20", "8")
```

```python
def average(a, b):
    try:
        result = (a + b) / 2
    except TypeError:
        print("Invalid input")
    else:
        print("Average:", result)
    finally:
        print("Average calculation done")


average(10, 20)
```

```python
import math

def square_root(value):
    try:
        num = float(value)
        result = math.sqrt(num)
    except ValueError:
        print("Invalid input")
    else:
        print("Square root:", result)
    finally:
        print("Square root operation finished")


square_root("25")
```

```python
def simple_interest(p, r, t):
    try:
        interest = (p * r * t) / 100
    except TypeError:
        print("Invalid input")
    else:
        print("Simple Interest:", interest)
    finally:
        print("Interest calculation attempt done")


simple_interest(10000, 5, 2)
```

```python
def remainder(a, b):
    try:
        num1 = int(a)
        num2 = int(b)
        result = num1 % num2
    except ValueError:
        print("Invalid input")
    except ZeroDivisionError:
        print("Cannot divide by zero")
    else:
        print("Remainder:", result)
    finally:
        print("Remainder operation finished")


remainder("17", "5")
```
