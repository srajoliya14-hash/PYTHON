```python
#  Create a function calculate_area(length, breadth) that returns the area of a rectangle. Pass length =
#15 and breadth = 8 and print the area.
# Function Definition
def calculate_area(length, breadth):
    return length * breadth

# Function Call
area = calculate_area(15, 8)

# Print Result
print("Area of Rectangle =", area)
```

```python
## Function Definition
 # Function Definition
def add(a, b):
    return a + b

def subtract(a, b):
    return a - b

# Function Calls
print("Addition =", add(20, 5))
print("Subtraction =", subtract(20, 5))
```

```python
## Function Definition
def cube(num):
    return num ** 3

def double(num):
    return num * 2

# Function Calls
print("Cube =", cube(4))
print("Double =", double(4))
```

```python
#Create functions square(num) and double(num). Find the square of the double of 5 using
#square(double(5)).
# Function Definition
def square(num):
    return num * num

def double(num):
    return num * 2

# Nested Function Call
print(square(double(5)))
```

```python
#Create functions add(a, b) and multiply(a, b). Find (add(5, 3)) × 2 using nested function calls.
 # Function Definition
def add(a, b):
    return a + b

def multiply(a, b):
    return a * b

# Nested Function Call
print(multiply(add(5, 3), 2))
```
