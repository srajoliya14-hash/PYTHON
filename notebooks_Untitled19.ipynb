```python
marks = []

for i in range(6):
    m = int(input(f"Enter marks of subject {i+1}: "))
    marks.append(m)

total = sum(marks)
average = total / 6
highest = max(marks)
lowest = min(marks)

if average >= 90:
    grade = "A+"
elif average >= 80:
    grade = "A"
elif average >= 70:
    grade = "B"
elif average >= 60:
    grade = "C"
elif average >= 50:
    grade = "D"
else:
    grade = "F"

print("Marks:", marks)
print("Total:", total)
print("Average:", average)
print("Highest:", highest)
print("Lowest:", lowest)
print("Grade:", grade)
```

```python
nums = [5,2,-3,0,4,8,-1,6,0,7,9,-5,12,14,1]

even = odd = positive = negative = zero = 0

for num in nums:
    if num % 2 == 0:
        even += 1
    else:
        odd += 1

    if num > 0:
        positive += 1
    elif num < 0:
        negative += 1
    else:
        zero += 1

print("Even:", even)
print("Odd:", odd)
print("Positive:", positive)
print("Negative:", negative)
print("Zero:", zero)
```

```python
arr = [4,6,2,5,4,8,6,9,10,2,3,7,5,8,12,13,6,4,15,16]

duplicates = []

for num in arr:
    if arr.count(num) > 1 and num not in duplicates:
        duplicates.append(num)

print("Duplicate values:", duplicates)
```

```python
def count_prime(numbers):
    count = 0

    for num in numbers:
        if num > 1:
            prime = True

            for i in range(2, int(num**0.5)+1):
                if num % i == 0:
                    prime = False
                    break

            if prime:
                count += 1

    return count


nums = [2,3,4,5,6,7,11,12]

print("Prime Count:", count_prime(nums))
```

```python
players = {
    "Rahul":45,
    "Aman":76,
    "Riya":32,
    "Neha":55,
    "Karan":18,
    "Priya":88,
    "Rohit":60,
    "Ankit":49
}

scores = list(players.values())

highest = max(players, key=players.get)
lowest = min(players, key=players.get)

print("Highest Scorer:", highest, players[highest])
print("Lowest Scorer:", lowest, players[lowest])
print("Total Score:", sum(scores))

print("Players scoring above 50:")

for name, score in players.items():
    if score > 50:
        print(name, score)
```
