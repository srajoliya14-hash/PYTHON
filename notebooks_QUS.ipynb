```python
#Write a function that checks whether a number is even or odd.
def check_even_odd(num):
    if num % 2 == 0:
        return "Even"
    else:
        return "Odd"

# Input from the user
n = int(input("Enter a number: "))

# Call the function and print the result
print(check_even_odd(n))
```

```python
#Write a function that takes a string and returns its length without using len().
def string_length(text):
    count = 0
    for ch in text:
        count += 1
    return count

# Input from the user
s = input("Enter a string: ")

# Call the function and print the result
print("Length =", string_length(s))
```

```python
#Write a function that returns the larger of two numbers.
def larger(a, b):
    if a > b:
        return a
    else:
        return b

# Input from the user
num1 = int(input("Enter first number: "))
num2 = int(input("Enter second number: "))

# Call the function and print the result
print("Larger number =", larger(num1, num2))
```

```python
#7. Write a function that accepts a list of numbers and returns the largest element.
def largest_element(lst):
    largest = lst[0]

    for num in lst:
        if num > largest:
            largest = num

    return largest

# Input from the user
numbers = list(map(int, input("Enter numbers separated by space: ").split()))

# Call the function and print the result
print("Largest element =", largest_element(numbers))
```

```python
#8. Write a function that calculates the factorial of a number using a loop.
def factorial(n):
    fact = 1
    

    for i in range(1, n + 1):
        fact = fact * i

    return fact

# Input from the user
num = int(input("Enter a number: "))



# Call the function and print the result
print("Factorial =", factorial(num))
```

```python
#9. Write a function that returns the reverse of a string.
def reverse_string(text):
    rev = ""

    for ch in text:
        rev = ch + rev

    return rev

# Input from the user
s = input("Enter a string: ")

# Call the function and print the result
print("Reversed string =", reverse_string(s))
```

```python
#10. Write a function that checks whether a given string is a palindrome.
def is_palindrome(text):
    rev = ""

    for ch in text:
        rev = ch + rev

    if text == rev:
        return "Palindrome"
    else:
        return "Not Palindrome"

# Input from the user
s = input("Enter a string: ")

# Call the function and print the result
print(is_palindrome(s))
```

```python
#11. Write a function that accepts a list and returns a new list containing only the even numbers.
def even_numbers(lst):
    even = []

    for num in lst:
        if num % 2 == 0:
            even.append(num)

    return even

# Input from the user
numbers = list(map(int, input("Enter numbers separated by space: ").split()))

# Call the function and print the result
print("Even numbers =", even_numbers(numbers))
```
