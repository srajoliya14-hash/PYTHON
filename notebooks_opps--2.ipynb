```python
class complex:
    def __init__(self,real,img):
        self.real = real
        self.img = img
    def shownumber(self):
        print(self.real,"i+",self.img,"j")
      #  print(self.real,"i+"self.img,"j")
    def __add__(num1,num2):
        newreal = num1.real + num2.real
        newimg = num1.img + num2.img
        return complex(newreal,newimg)
                      
num1 = complex(1,3)
num1.shownumber()
num2 = complex(2,4)
num2.shownumber() 
num3 = num1 + num2
num3.shownumber()
```

```python
class circle:
    def __init__(self,radius):
        self.radius = radius
    def area(self):
        return 3.14*self.radius**2
    def peramitar(self):
        return 2*3.14*self.radius
        
c1 = circle(21)    
print(c1.area,c1.peramitar)
```
